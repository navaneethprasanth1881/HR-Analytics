```python
dep_left=df.groupby("dept")["left"].sum().sort_values(ascending=False)
dep_left
```



*  1014 employee have left from sales department
*   followed by technical department with 697 employees and support department with 555 employees



```python
plt.figure(figsize=(12,15))
plt.bar(dep_left.index,dep_left.values)
plt.ylabel("left")
plt.xlabel("department")
plt.show()
```

```python
proj_left=df.groupby("numberOfProjects")["left"].sum()
proj_left
```

```python
plt.figure(figsize=(12,15))
plt.bar(proj_left.index,proj_left.values)
plt.ylabel("left")
plt.xlabel("No fo projects")
plt.show()
```



*  employees with 2 projects has left the most
*   followed by 5,6,4,7

*  Employees with 3 projects have left least



###Insights
That means employees with less than 3 projects left the more but the no of projects does not directly affect the employees who have left





```python
sal_left=df.groupby("salary")["left"].sum().sort_values(ascending=False)
sal_left
```

```python
plt.bar(sal_left.index,sal_left.values)
plt.xlabel("Salary")
plt.ylabel("Left")
plt.show()
```

####Insights
Employees with low salary has more tendencies to left and high salary employees has lower tendency to left.From this we have understood that salary affects the employees left

```python
dep_rate=df.groupby("dept")["left"].mean().sort_values(ascending=False)
dep_rate
```

```python
plt.figure(figsize=(12,15))
plt.bar(dep_rate.index,dep_rate.values)
plt.ylabel("left")
plt.xlabel("department")
plt.show()
```

####Insights
We found out that hr has the highest rate of leaving the company
